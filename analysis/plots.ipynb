```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv('log6.csv')

# Create subplots
fig, axs = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Motor Performance Characteristics (Line Graphs)', fontsize=16)

# We should sort the data by the x-axis for each plot to ensure smooth lines
# Plot 1: v_mot vs m_rpm
df_sorted = df.sort_values(by='m_rpm')
axs[0, 0].plot(df_sorted['m_rpm'], df_sorted['v_mot'], color='blue', marker='o', alpha=0.7)
axs[0, 0].set_title('v_mot vs m_rpm')
axs[0, 0].set_xlabel('m_rpm')
axs[0, 0].set_ylabel('v_mot (V)')
axs[0, 0].grid(True)

# Plot 2: v_mot vs o_rpm
df_sorted = df.sort_values(by='o_rpm')
axs[0, 1].plot(df_sorted['o_rpm'], df_sorted['v_mot'], color='green', marker='o', alpha=0.7)
axs[0, 1].set_title('v_mot vs o_rpm')
axs[0, 1].set_xlabel('o_rpm')
axs[0, 1].set_ylabel('v_mot (V)')
axs[0, 1].grid(True)

# Plot 3: i_mot vs m_rpm
df_sorted = df.sort_values(by='m_rpm')
axs[0, 2].plot(df_sorted['m_rpm'], df_sorted['i_mot'], color='red', marker='o', alpha=0.7)
axs[0, 2].set_title('i_mot vs m_rpm')
axs[0, 2].set_xlabel('m_rpm')
axs[0, 2].set_ylabel('i_mot (A)')
axs[0, 2].grid(True)

# Plot 4: i_mot vs o_rpm
df_sorted = df.sort_values(by='o_rpm')
axs[1, 0].plot(df_sorted['o_rpm'], df_sorted['i_mot'], color='orange', marker='o', alpha=0.7)
axs[1, 0].set_title('i_mot vs o_rpm')
axs[1, 0].set_xlabel('o_rpm')
axs[1, 0].set_ylabel('i_mot (A)')
axs[1, 0].grid(True)

# Plot 5: i_mot vs v_mot (Changed from v_mot vs i_mot)
df_sorted = df.sort_values(by='v_mot')
axs[1, 1].plot(df_sorted['v_mot'], df_sorted['i_mot'], color='purple', marker='o', alpha=0.7)
axs[1, 1].set_title('i_mot vs v_mot')
axs[1, 1].set_xlabel('v_mot (V)')
axs[1, 1].set_ylabel('i_mot (A)')
axs[1, 1].grid(True)

# Remove empty subplot
fig.delaxes(axs[1, 2])

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('motor_line_plots_log6.png', dpi=300)
print("Plots generated and saved as motor_line_plots_log6.png")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

def overlay_motor_plots(csv_filenames):
    # Set up the plotting grid (2 rows, 3 columns)
    fig, axs = plt.subplots(2, 3, figsize=(16, 10))
    fig.canvas.manager.set_window_title('Motor Data Comparison')
    fig.suptitle('Motor Performance Characteristics Comparison', fontsize=16)

    # Loop through each file in the list
    for file in csv_filenames:
        if not os.path.exists(file):
            print(f"Warning: Could not find '{file}'. Skipping.")
            continue

        print(f"Processing {file}...")
        df = pd.read_csv(file)
        
        # Use the filename for the legend label
        label_name = os.path.basename(file)

        # Plot 1: v_mot vs m_rpm
        df_sorted = df.sort_values(by='m_rpm')
        axs[0, 0].plot(df_sorted['m_rpm'], df_sorted['v_mot'], marker='o', alpha=0.7, label=label_name)

        # Plot 2: v_mot vs o_rpm
        df_sorted = df.sort_values(by='o_rpm')
        axs[0, 1].plot(df_sorted['o_rpm'], df_sorted['v_mot'], marker='o', alpha=0.7, label=label_name)

        # Plot 3: i_mot vs m_rpm
        df_sorted = df.sort_values(by='m_rpm')
        axs[0, 2].plot(df_sorted['m_rpm'], df_sorted['i_mot'], marker='o', alpha=0.7, label=label_name)

        # Plot 4: i_mot vs o_rpm
        df_sorted = df.sort_values(by='o_rpm')
        axs[1, 0].plot(df_sorted['o_rpm'], df_sorted['i_mot'], marker='o', alpha=0.7, label=label_name)

        # Plot 5: i_mot vs v_mot
        df_sorted = df.sort_values(by='v_mot')
        axs[1, 1].plot(df_sorted['v_mot'], df_sorted['i_mot'], marker='o', alpha=0.7, label=label_name)

    # --- Formatting all plots ---
    
    # 1. v_mot vs m_rpm
    axs[0, 0].set_title('v_mot vs m_rpm')
    axs[0, 0].set_xlabel('m_rpm')
    axs[0, 0].set_ylabel('v_mot (V)')
    axs[0, 0].grid(True, linestyle='--', alpha=0.6)
    axs[0, 0].legend()

    # 2. v_mot vs o_rpm
    axs[0, 1].set_title('v_mot vs o_rpm')
    axs[0, 1].set_xlabel('o_rpm')
    axs[0, 1].set_ylabel('v_mot (V)')
    axs[0, 1].grid(True, linestyle='--', alpha=0.6)
    axs[0, 1].legend()

    # 3. i_mot vs m_rpm
    axs[0, 2].set_title('i_mot vs m_rpm')
    axs[0, 2].set_xlabel('m_rpm')
    axs[0, 2].set_ylabel('i_mot (A)')
    axs[0, 2].grid(True, linestyle='--', alpha=0.6)
    axs[0, 2].legend()

    # 4. i_mot vs o_rpm
    axs[1, 0].set_title('i_mot vs o_rpm')
    axs[1, 0].set_xlabel('o_rpm')
    axs[1, 0].set_ylabel('i_mot (A)')
    axs[1, 0].grid(True, linestyle='--', alpha=0.6)
    axs[1, 0].legend()

    # 5. i_mot vs v_mot
    axs[1, 1].set_title('i_mot vs v_mot')
    axs[1, 1].set_xlabel('v_mot (V)')
    axs[1, 1].set_ylabel('i_mot (A)')
    axs[1, 1].grid(True, linestyle='--', alpha=0.6)
    axs[1, 1].legend()

    # Remove the empty 6th subplot (bottom right)
    fig.delaxes(axs[1, 2])

    # Adjust layout so labels don't overlap
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    
    # Save the figure as a PNG file
    output_filename = 'motor_comparison_plots.png'
    plt.savefig(output_filename, dpi=300)
    print(f"\nSuccess! Comparison plots saved to '{output_filename}'")

    # Display the plots in a window
    plt.show()

if __name__ == "__main__":
    # ---> UPDATE THIS LIST WITH YOUR ACTUAL CSV FILE NAMES <---
    # You can add as many as you want separated by commas!
    my_csv_files = [ 
        'log5.csv',
        'log6.csv',
        'log8.csv'
    ] 
    
    overlay_motor_plots(my_csv_files)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

def overlay_all_plots(csv_filenames):
    # Set up the plotting grid (3 rows, 3 columns = 9 plots total)
    fig, axs = plt.subplots(3, 3, figsize=(18, 15))
    fig.canvas.manager.set_window_title('Comprehensive Motor Data Comparison')
    fig.suptitle('Motor Performance & Efficiency Characteristics', fontsize=18)

    # Loop through each file in the list
    for file in csv_filenames:
        if not os.path.exists(file):
            print(f"Warning: Could not find '{file}'. Skipping.")
            continue

        print(f"Processing {file}...")
        df = pd.read_csv(file)
        
        # Use the filename for the legend label
        label_name = os.path.basename(file)

        # Plot 1: v_mot vs m_rpm
        df_sorted = df.sort_values(by='m_rpm')
        axs[0, 0].plot(df_sorted['m_rpm'], df_sorted['v_mot'], marker='o', alpha=0.7, label=label_name)

        # Plot 2: v_mot vs o_rpm
        df_sorted = df.sort_values(by='o_rpm')
        axs[0, 1].plot(df_sorted['o_rpm'], df_sorted['v_mot'], marker='o', alpha=0.7, label=label_name)

        # Plot 3: i_mot vs m_rpm
        df_sorted = df.sort_values(by='m_rpm')
        axs[0, 2].plot(df_sorted['m_rpm'], df_sorted['i_mot'], marker='o', alpha=0.7, label=label_name)

        # Plot 4: i_mot vs o_rpm
        df_sorted = df.sort_values(by='o_rpm')
        axs[1, 0].plot(df_sorted['o_rpm'], df_sorted['i_mot'], marker='o', alpha=0.7, label=label_name)

        # Plot 5: i_mot vs v_mot
        df_sorted = df.sort_values(by='v_mot')
        axs[1, 1].plot(df_sorted['v_mot'], df_sorted['i_mot'], marker='o', alpha=0.7, label=label_name)

        # Plot 6: eff vs v_mot
        df_sorted = df.sort_values(by='v_mot')
        axs[1, 2].plot(df_sorted['v_mot'], df_sorted['eff'], marker='o', alpha=0.7, label=label_name)

        # Plot 7: eff vs i_mot
        df_sorted = df.sort_values(by='i_mot')
        axs[2, 0].plot(df_sorted['i_mot'], df_sorted['eff'], marker='o', alpha=0.7, label=label_name)

        # Plot 8: eff vs o_rpm
        df_sorted = df.sort_values(by='o_rpm')
        axs[2, 1].plot(df_sorted['o_rpm'], df_sorted['eff'], marker='o', alpha=0.7, label=label_name)


    # --- Formatting all plots ---
    plots_info = [
        (axs[0, 0], 'v_mot vs m_rpm', 'm_rpm', 'v_mot (V)'),
        (axs[0, 1], 'v_mot vs o_rpm', 'o_rpm', 'v_mot (V)'),
        (axs[0, 2], 'i_mot vs m_rpm', 'm_rpm', 'i_mot (A)'),
        (axs[1, 0], 'i_mot vs o_rpm', 'o_rpm', 'i_mot (A)'),
        (axs[1, 1], 'i_mot vs v_mot', 'v_mot (V)', 'i_mot (A)'),
        (axs[1, 2], 'Efficiency vs Motor Voltage', 'v_mot (V)', 'Efficiency (%)'),
        (axs[2, 0], 'Efficiency vs Motor Current', 'i_mot (A)', 'Efficiency (%)'),
        (axs[2, 1], 'Efficiency vs Output RPM', 'o_rpm', 'Efficiency (%)')
    ]

    for ax, title, xlabel, ylabel in plots_info:
        ax.set_title(title)
        ax.set_xlabel(xlabel)
        ax.set_ylabel(ylabel)
        ax.grid(True, linestyle='--', alpha=0.6)
        ax.legend()

    # Remove the empty 9th subplot (bottom right)
    fig.delaxes(axs[2, 2])

    # Adjust layout so labels don't overlap
    plt.tight_layout(rect=[0, 0.03, 1, 0.96])
    
    # Save the figure as a PNG file
    output_filename = 'comprehensive_motor_plots.png'
    plt.savefig(output_filename, dpi=300)
    print(f"\nSuccess! Comprehensive plots saved to '{output_filename}'")


my_csv_files = [ 
    'log5.csv',
    'log6.csv',
    'log8.csv'
] 

overlay_all_plots(my_csv_files)
```
